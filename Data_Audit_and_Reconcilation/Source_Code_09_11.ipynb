```python
# Correlation of attributes in Frequency data
corr_matrix = frequency_data.corr(numeric_only=True)

plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, annot=True, vmin=-1, vmax=1.0
            
            , fmt='.2f')
plt.title('Correlation Matrix Heatmap')
plt.show()
```
