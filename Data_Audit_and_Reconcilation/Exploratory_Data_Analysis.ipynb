```python
final_severity_data = (
    severity_data.groupby('IDpol', as_index=False)
    .agg(
        Claim_Count=('ClaimAmount', 'size'),
        Total_Claim_Amount=('ClaimAmount', 'sum'),
    )
)
```

```python
final_severity_data = pd.DataFrame(final_severity_data)
```

```python
final_severity_data
```

```python
joined_data = frequency_data.merge(final_severity_data, on='IDpol', how='left', validate='one_to_many')
```
